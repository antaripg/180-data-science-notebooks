### Common Array Creation Methods

```python

# Array filled with zeros
zeros = np.zeros(5)

# Array filled with ones
ones = np.ones((2, 3))

# Sequence with a fixed step
sequence = np.arange(0, 10, 2)

# Fixed number of evenly spaced values
evenly_spaced = np.linspace(0, 1, 5)

# Identity matrix
identity = np.eye(3)

print("Zeros:", zeros)
print("\nOnes:\n", ones)
print("\nSequence:", sequence)
print("\nEvenly spaced:", evenly_spaced)
print("\nIdentity matrix:\n", identity)
```

### Dimensions, shape, size and dtype

```python

one_d = np.array([10, 20, 30, 40])

two_d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

three_d = np.arange(24).reshape(2, 3, 4)

arrays = {
    "one_d": one_d,
    "two_d": two_d,
    "three_d": three_d
}

for name, array in arrays.items():
    print(f"\n{name}")
    print("Array:\n", array)
    print("Dimensions:", array.ndim)
    print("Shape:", array.shape)
    print("Total elements:", array.size)
    print("Data type:", array.dtype)
```

Key distinction:
- ndim tells you the number of dimensions.
- shape tells you the length of each dimension.
- size tells you the total number of elements.
- dtype tells you the element data type.

### Understanding Data Types

```python

integers = np.array([1, 2, 3])
decimals = np.array([1.5, 2.5, 3.5])

float32_array = np.array([1, 2, 3], dtype=np.float32)
int32_array = np.array([1.8, 2.4, 3.9]).astype(np.int32)

print("Integer dtype:", integers.dtype)
print("Decimal dtype:", decimals.dtype)
print("Float32 dtype:", float32_array.dtype)

# Conversion truncates fractional parts when converting to integers
print("Converted to integers:", int32_array)
```

### Indexing And Slicing

```python

arr = np.array([10, 20, 30, 40, 50, 60])

print("First element:", arr[0])
print("Last element:", arr[-1])
print("Elements from index 1 to 3:", arr[1:4])
print("Every second element:", arr[::2])
print("Reversed array:", arr[::-1])
```

### Indexing Two Dimensional Array

```python

matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Element at row 1, column 2:", matrix[1, 2])

print("\nFirst row:", matrix[0, :])

print("\nSecond column:", matrix[:, 1])

print("\nTop-left 2x2 block:\n", matrix[:2, :2])
```

### Views Vs. Copies

```python

original = np.array([1, 2, 3, 4])

# Basic slicing typically returns a view
view = original[1:3]
view[0] = 200

print("Original after modifying view:", original)

# copy() creates an independent array
original = np.array([1, 2, 3, 4])

independent = original[1:3].copy()
independent[0] = 200

print("Original after modifying copy:", original)
print("Independent array:", independent)
```

### Boolean Indexing

```python

temperatures = np.array([72.5, 81.0, 95.2, 68.4, 102.1, 88.0])

mask = temperatures > 90

print("Boolean mask:", mask)
print("Temperatures above 90:", temperatures[mask])

# Combine conditions with & and parentheses
normal_range = temperatures[
    (temperatures >= 70) & (temperatures <= 95)
]

print("Temperatures between 70 and 95:", normal_range)
```

### Reshaping and Transposing

```python

values = np.arange(1, 13)

matrix = values.reshape(3, 4)

print("Original array:", values)
print("\nReshaped array:\n", matrix)
print("\nTranspose:\n", matrix.T)
print("\nFlattened copy:", matrix.flatten())
print("\nRaveled array:", matrix.ravel())
```

reshape() changes the array's shape without changing the number of elements. flatten() returns a copy, whereas ravel() returns a view when possible.

### Vectorized Arithmetic

```python

a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("Addition:", a + b)
print("Subtraction:", b - a)
print("Elementwise multiplication:", a * b)
print("Division:", b / a)
print("Squared:", a ** 2)

print("Sum:", a.sum())
print("Mean:", a.mean())
print("Minimum:", a.min())
print("Maximum:", a.max())
```

### Vectorization Vs. Loops

```python

from timeit import timeit

values = np.arange(1, 1_000_001)

def square_with_loop(array):
    result = []
    for value in array:
        result.append(value * value)
    return result

def square_vectorized(array):
    return array * array

# Verify that the results agree
sample = np.array([1, 2, 3, 4])

assert square_with_loop(sample) == square_vectorized(sample).tolist()

# Benchmark a smaller array for a quick experiment
sample_data = values[:10_000]

loop_time = timeit(
    lambda: square_with_loop(sample_data),
    number=3
)

vectorized_time = timeit(
    lambda: square_vectorized(sample_data),
    number=3
)

print(f"Loop time:       {loop_time:.6f} seconds")
print(f"Vectorized time: {vectorized_time:.6f} seconds")
```

Vectorization usually reduces Python-level overhead for numerical operations. The exact speed-up depends on the operation, array size, and hardware, so measure rather than assume.

### Broadcasting

```python

# Broadcasting a scalar
measurements = np.array([10, 20, 30])

print(measurements + 5)

# Broadcasting a 1D array across matrix columns
data = np.array([
    [10, 100, 1000],
    [20, 200, 2000]
])

offsets = np.array([1, 10, 100])

print("\nOriginal data:\n", data)
print("\nOffsets:", offsets)
print("\nResult:\n", data + offsets)
```

### Aggregation along Axes

```python

data = np.array([
    [10.0, 100.0],
    [20.0, 200.0],
    [30.0, 300.0]
])

print("Column means:", data.mean(axis=0))
print("Row means:", data.mean(axis=1))

column_means = data.mean(axis=0)
centered = data - column_means

print("\nColumn-centered data:\n", centered)
print("New column means:", centered.mean(axis=0))
```

For a two-dimensional array:
- axis=0 aggregates down the rows, producing one value per column.
- axis=1 aggregates across the columns, producing one value per row.

### Mini Project: Industrial Sensor Analysis

This mini project brings together array creation, shapes, vectorized calculations, and Boolean Filtering

```python

# Rows are time steps; columns are sensors.
readings = np.array([
    [21.0, 45.0, 101.0],
    [22.5, 47.0, 99.0],
    [23.0, 110.0, 102.0],
    [20.5, 46.0, 100.0]
])

sensor_names = ["temperature", "pressure", "vibration"]

print("Shape:", readings.shape)
print("Time steps:", readings.shape[0])
print("Sensors:", readings.shape[1])

print("\nAverage per sensor:")
for name, value in zip(sensor_names, readings.mean(axis=0)):
    print(f"{name}: {value:.2f}")

print("\nMaximum per sensor:")
for name, value in zip(sensor_names, readings.max(axis=0)):
    print(f"{name}: {value:.2f}")
```

### Detect Threshold Exceedances

```python

thresholds = np.array([23.0, 80.0, 105.0])

# Compare every sensor reading with its corresponding threshold.
exceeds_threshold = readings > thresholds

# Flag a time step if any sensor exceeds its threshold.
flagged_rows = exceeds_threshold.any(axis=1)

print("Threshold exceedances:\n", exceeds_threshold)
print("Flagged time-step indices:", np.where(flagged_rows)[0])
print("Flagged readings:\n", readings[flagged_rows])
```

This is a demonstration of array operations, not a validated fault-detection system. Real industrial monitoring needs domain-validated thresholds and checks for sensor quality and operating conditions.
